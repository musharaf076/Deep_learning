```python
import numpy as np
def step_function(x):
  return np.where(x>=0,1,0)
def predict(X, weights, bias):
  linear_output = np.dot(X,weights) + bias
  return step_function(linear_output)
def train_perceptron(X, y, learning_rate=0.01, epochs=20):
  input_size = X.shape[1]
  weights = np.zeros(input_size)
  bias = 0
  for epoch in range(epochs):
    total_error = 0
    for i in range(len(X)):
      current_prediction = predict(X[i], weights , bias)
      error = y[i]-current_prediction

      weights+=learning_rate * error *X[i]
      bias += learning_rate * error
      total_error += abs(error)
    if total_error == 0:
      print(f"Converged at epoch{epoch +1}")
      break
  return weights, bias
def test_gate(gate_name, X,y):
  print(f"\n================={gate_name}================ \n")
  weights, bias = train_perceptron(X, y , learning_rate=0.1,epochs=20)
  print(f"final weights: {weights}")
  print(f"final bias: {bias:.2f}")

  print("\ntest result")
  for inputs, target in zip(X,y):
    pred = predict(inputs, weights, bias)
    print(f"input: {inputs}, target: {target}, prediction: {pred}")
```

```python
X_and = np.array([[0,0],[0,1],[1,0],[1,1]])
y_and = np.array([0,0,0,1])
X_or =  np.array([[0,0],[0,1],[1,0],[1,1]])
y_or = np.array([0,1,1,1])
X_not = np.array([[0],[1]])
y_not = np.array([1,0])
```

```python
if __name__ == "__main__":
  test_gate("AND gate", X_and, y_and)
  test_gate("OR gate", X_or, y_or)
  test_gate("NOT gate", X_not, y_not)
```
