```python
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam, SGD

from sklearn.datasets import fetch_california_housing, load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

import keras_tuner as kt
import pandas as pd
import numpy as np
```

```python

# Reproducibility
tf.random.set_seed(42)
np.random.seed(42)

# ==========================================
# PART A: REGRESSION (California Housing)
# ==========================================

print("="*50)
print("Part A: REGRESSION (California Dataset)")
print("="*50)

# Load dataset
housing = fetch_california_housing()

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    housing.data,
    housing.target,
    test_size=0.2,
    random_state=42
)

# Feature scaling
scaler = StandardScaler()

X_train_r_scaled = scaler.fit_transform(X_train_r)
X_test_r_scaled = scaler.transform(X_test_r)

# Build regression model
reg_model = Sequential([
    Dense(64, activation="relu",
          input_shape=(X_train_r_scaled.shape[1],)),
    Dense(32, activation="relu"),
    Dense(1)
])

# Compile model
reg_model.compile(
    optimizer="adam",
    loss="mean_squared_error",
    metrics=["mae"]
)

# Train model
history_reg = reg_model.fit(
    X_train_r_scaled,
    y_train_r,
    epochs=25,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)

# Evaluate model
test_mse, test_mae = reg_model.evaluate(
    X_test_r_scaled,
    y_test_r,
    verbose=0
)

print("Test MSE:", test_mse)
print("Test MAE:", test_mae)


# ==========================================
# PART B: MULTICLASS CLASSIFICATION (Iris)
# ==========================================

print("="*50)
print("Part B: MULTICLASS CLASSIFICATION (Iris Dataset)")
print("="*50)

# Load dataset
iris = load_iris()

X_c_train, X_c_test, y_c_train, y_c_test = train_test_split(
    iris.data,
    iris.target,
    test_size=0.2,
    random_state=42,
    stratify=iris.target
)

# Feature scaling
scaler_c = StandardScaler()

X_c_train_scaled = scaler_c.fit_transform(X_c_train)
X_c_test_scaled = scaler_c.transform(X_c_test)

# Build classification model
clf_model = Sequential([
    Dense(64, activation="relu",
          input_shape=(X_c_train_scaled.shape[1],)),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(3, activation="softmax")
])

# Compile model
clf_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train model
history_clf = clf_model.fit(
    X_c_train_scaled,
    y_c_train,
    epochs=25,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)

# Evaluate model
test_loss, test_accuracy = clf_model.evaluate(
    X_c_test_scaled,
    y_c_test,
    verbose=0
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```
