```python


import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, GRU, Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import confusion_matrix, classification_report

# 1. LOAD IMDB DATASET
vocab_size = 10000
max_length = 200

(X_train, y_train), (X_test, y_test) = imdb.load_data(
    num_words=vocab_size
)

print("Training Samples:", X_train.shape)
print("Testing Samples:", X_test.shape)
print("First Review:", X_train[0])
print("Sentiment Label:", y_train[0])

# 2. PAD SEQUENCES
X_train = pad_sequences(
    X_train, maxlen=max_length,
    padding="post", truncating="post"
)

X_test = pad_sequences(
    X_test, maxlen=max_length,
    padding="post", truncating="post"
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)

# 3. DECODE FIRST REVIEW
word_index = imdb.get_word_index()

reverse_word_index = {
    value + 3: key for key, value in word_index.items()
}

reverse_word_index[0] = "<PAD>"
reverse_word_index[1] = "<START>"
reverse_word_index[2] = "<UNK>"
reverse_word_index[3] = "<UNUSED>"

decoded_review = " ".join(
    reverse_word_index.get(i, "?") for i in X_train[0]
)

print("\nDecoded Review:\n", decoded_review)
print("Sentiment:", "Positive" if y_train[0] == 1 else "Negative")

# 4. BUILD LSTM MODEL
embedding_dim = 128

lstm_model = Sequential([
    Input(shape=(max_length,)),
    Embedding(vocab_size, embedding_dim),
    LSTM(128),
    Dropout(0.5),
    Dense(64, activation="relu"),
    Dense(1, activation="sigmoid")
])

lstm_model.summary()

# 5. COMPILE LSTM
lstm_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# 6. TRAIN LSTM
history_lstm = lstm_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=5,
    batch_size=64,
    verbose=1
)

# 7. EVALUATE LSTM
loss_lstm, accuracy_lstm = lstm_model.evaluate(
    X_test, y_test, verbose=0
)

print("\nLSTM Test Loss:", loss_lstm)
print("LSTM Test Accuracy:", accuracy_lstm)

# 8. BUILD GRU MODEL
gru_model = Sequential([
    Input(shape=(max_length,)),
    Embedding(vocab_size, embedding_dim),
    GRU(128),
    Dropout(0.5),
    Dense(64, activation="relu"),
    Dense(1, activation="sigmoid")
])

gru_model.summary()

# 9. COMPILE GRU
gru_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# 10. EARLY STOPPING
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)

# 11. TRAIN GRU
history_gru = gru_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=10,
    batch_size=64,
    callbacks=[early_stop],
    verbose=1
)

# 12. EVALUATE GRU
loss_gru, accuracy_gru = gru_model.evaluate(
    X_test, y_test, verbose=0
)

print("\nGRU Test Loss:", loss_gru)
print("GRU Test Accuracy:", accuracy_gru)

# 13. COMPARE MODELS
comparison = pd.DataFrame({
    "Model": ["LSTM", "GRU"],
    "Loss": [loss_lstm, loss_gru],
    "Accuracy": [accuracy_lstm, accuracy_gru]
})

print("\nMODEL COMPARISON")
print(comparison)

# 14. MAKE PREDICTIONS
y_pred_lstm = (lstm_model.predict(X_test) > 0.5).astype(int)
y_pred_gru = (gru_model.predict(X_test) > 0.5).astype(int)

print("\nLSTM Prediction Shape:", y_pred_lstm.shape)
print("GRU Prediction Shape:", y_pred_gru.shape)

# 15. LSTM CONFUSION MATRIX
cm_lstm = confusion_matrix(y_test, y_pred_lstm)

plt.figure(figsize=(6, 6))
plt.imshow(cm_lstm, cmap="Blues")
plt.title("LSTM Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()
plt.show()

# 16. GRU CONFUSION MATRIX
cm_gru = confusion_matrix(y_test, y_pred_gru)

plt.figure(figsize=(6, 6))
plt.imshow(cm_gru, cmap="Greens")
plt.title("GRU Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()
plt.show()

# 17. CLASSIFICATION REPORT
print("\n LSTM ")
print(classification_report(
    y_test,
    y_pred_lstm,
    target_names=["Negative", "Positive"]
))

print("\n GRU" )
print(classification_report(
    y_test,
    y_pred_gru,
    target_names=["Negative", "Positive"]
))

# 18. ACCURACY COMPARISON GRAPH
plt.figure(figsize=(8, 5))

plt.plot(history_lstm.history["accuracy"], label="LSTM Train", color="#FFA500")
plt.plot(history_lstm.history["val_accuracy"], label="LSTM Validation", color="#FFA500", linestyle="--")
plt.plot(history_gru.history["accuracy"], label="GRU Train", color="dodgerblue")
plt.plot(history_gru.history["val_accuracy"], label="GRU Validation", color="dodgerblue", linestyle="--")
for i in range(2):
    for j in range(2):
        plt.text(j, i, cm_lstm[i, j], ha="center", va="center")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("LSTM vs GRU Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# 19. LOSS COMPARISON GRAPH
plt.figure(figsize=(8, 5))

plt.plot(history_lstm.history["loss"], label="LSTM Train")
plt.plot(history_lstm.history["val_loss"], label="LSTM Validation")
plt.plot(history_gru.history["loss"], label="GRU Train")
plt.plot(history_gru.history["val_loss"], label="GRU Validation")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("LSTM vs GRU Loss")
plt.legend()
plt.grid(True)
plt.show()

```
